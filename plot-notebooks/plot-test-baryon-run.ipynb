# The second run for test
THe first run was using TATT model for 2pt.
Now we use NLA model consistently to see the improvement in the IA parameter constraint.

```python
def get(fname):
    chain, params, labels = utils.read_des_chain(fname, take=['om', 's8', 'sig8', 'a1', 'a2', 'alpha1', 'alpha2'])
    samples = utils.to_mcsamples(chain, params, labels)
    # utils.wplot(chain, params, labels)
    return samples
```

```python
# DM only
samples_2pt_map3_auto_334_344 = get('../output/test-second-run/des-y3-2pt-map3-addcross334-344.txt')
samples_2pt = get('../output/test-second-run/des-y3-2pt.txt')
# baryon
samples_2pt_map3_auto_334_344_baryon = get('../output/test-baryon-run/des-y3-2pt-map3-addcross334-344-NLA.txt')
samples_2pt_baryon = get('../output/test-baryon-run/des-y3-2pt.txt')
```

```python
samples_list = [samples_2pt, samples_2pt_baryon]
g = plots.get_single_plotter(ratio=1, width_inch=6)
g.triangle_plot(samples_list, ['om', 's8', 'sig8'], filled=False, \
    legend_labels=['2pt', '2pt (baryon)'], legend_loc='upper right', 
    markers={'om': 0.3, 's8': 0.8256, 'sig8': 0.8256, 'a1':0.7, 'alpha1':-1.7})
plt.show()

samples_list = [samples_2pt_map3_auto_334_344, samples_2pt_map3_auto_334_344_baryon]
g = plots.get_single_plotter(ratio=1, width_inch=6)
g.triangle_plot(samples_list, ['om', 's8', 'sig8'], filled=False, \
    legend_labels=['2pt+map3', '2pt+map3 (baryon)'], legend_loc='upper right', 
    markers={'om': 0.3, 's8': 0.8256, 'sig8': 0.8256, 'a1':0.7, 'alpha1':-1.7})
plt.show()
```

```python
names = ['s8', 'om']
samples_list = [samples_2pt, samples_2pt_map3_auto_334_344]
fom = np.array([utils.FoM_from_samples_names(s, names) for s in samples_list])
improve = fom/fom[0]
```
