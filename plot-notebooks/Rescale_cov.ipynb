```python
from astropy.io import fits

# Open the FITS file
input_filename = '../data/dv/sim_map3-NLA-cosmoDESY3.fits'
output_filename = '../data/dv/rescaled_sim_map3-NLA-cosmoDESY3.fits'

# Open the file and access the HDU list
with fits.open(input_filename) as hdul:
    # Access HDU 2
    hdu2 = hdul[2]
    
    # Print the header of HDU 2
    print(hdu2.header)
    
    # Get the data from HDU 2
    data = hdu2.data
    
    # Multiply the data by 0.31
    rescaled_data = data * 0.31
    
    # Update the HDU with the rescaled data
    hdu2.data = rescaled_data
    
    # Write to a new FITS file, preserving the original structure
    hdul.writeto(output_filename)
    
print(f'Rescaled data saved to {output_filename}')
```

```python
with fits.open(input_filename) as original_hdul, fits.open(output_filename) as rescaled_hdul:
    original_data = original_hdul[2].data
    rescaled_data = rescaled_hdul[2].data
    
    # Print data from the original file
    print("\nOriginal HDU[2] Data after loading the file again:")
    print(original_data)
    
    # Print data from the rescaled file
    print("\nRescaled HDU[2] Data:")
    print(rescaled_data)
```
