```python
import numpy as np
import matplotlib.pyplot as plt
from getdist import plots, MCSamples
import sys
sys.path.append('../python')
import utils
```

# Reading chains

```python
# No compression
s_2pt = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-run/des-y3-2pt.txt', to_mcsamples=True)
s_map3 = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-emu-run/des-y3-map3-all.txt', to_mcsamples=True)
s_2pt_map3 = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-emu-run/des-y3-2pt-map3-all.txt', to_mcsamples=True)
s_2pt_map3_auto = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-emu-run/des-y3-2pt-map3-auto.txt', to_mcsamples=True)
s_2pt_map3_auto_334_344 = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-emu-run/des-y3-2pt-map3-auto-334-344.txt', to_mcsamples=True)
# moped version
s_2pt_moped = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-moped-run/des-y3-2pt.txt', to_mcsamples=True)
s_map3_moped = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-moped-run/des-y3-map3-all.txt', to_mcsamples=True)
s_2pt_map3_moped = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-moped-run/des-y3-2pt-map3-all.txt', to_mcsamples=True)
s_2pt_map3_moped_auto = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-moped-run/des-y3-2pt-map3-auto.txt', to_mcsamples=True)
s_2pt_map3_moped_auto_334_344 = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-moped-run/des-y3-2pt-map3-auto-334-344.txt', to_mcsamples=True)
```

## Plotting
### MOPED correspondances for 2pt, map3, 2pt+map3

```python
g = plots.get_single_plotter(ratio=1, width_inch=4)
g.triangle_plot([s_2pt, s_2pt_moped], ['om', 's8', 'a1'], filled=False, \
    legend_labels=['2pt (analytic cov)', '2pt (analytic cov) + moped'], legend_loc='upper right', 
    markers={'om': 0.29, 's8':0.759, 'sig8': 0.772, 'a1':0.7})
plt.show()
```

```python
g = plots.get_single_plotter(ratio=1, width_inch=4)
g.triangle_plot([s_map3, s_map3_moped], ['om', 's8', 'a1'], filled=False, \
    legend_labels=['map3 (sim cov)', 'map3 (sim cov) + moped'], legend_loc='upper right', 
    markers={'om': 0.29, 's8':0.759, 'sig8': 0.772, 'a1':0.7})
plt.show()
```

```python
g = plots.get_single_plotter(ratio=1, width_inch=4)
g.triangle_plot([s_2pt_map3, s_2pt_map3_moped], ['om', 's8', 'a1'], filled=False, \
    legend_labels=['2pt(a)+map3(s)', '2pt(a)+map3(s) + moped'], legend_loc='upper right', 
    markers={'om': 0.29, 's8':0.759, 'sig8': 0.772, 'a1':0.7})
plt.show()
```

```python
g = plots.get_single_plotter(ratio=1, width_inch=4)
g.triangle_plot([s_2pt_map3_auto, s_2pt_map3_moped_auto], ['om', 's8', 'a1'], filled=False, \
    legend_labels=['2pt(a)+map3(s) [auto]', '2pt(a)+map3(s) [auto] + moped'], legend_loc='upper right', 
    markers={'om': 0.29, 's8':0.759, 'sig8': 0.772, 'a1':0.7})
plt.show()
```
