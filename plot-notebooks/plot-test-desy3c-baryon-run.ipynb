# Baryon
## Reading chains

```python
# No Baryon
s_joint_moped = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-joint-moped-run/des-y3-2pt-map3-all.txt', to_mcsamples=True)
s_joint_moped_auto = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-joint-moped-run/des-y3-2pt-map3-auto.txt', to_mcsamples=True)
s_joint_moped_auto_334_344 = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-joint-moped-run/des-y3-2pt-map3-auto-334-344.txt', to_mcsamples=True)
# Baryon
s_joint_moped_bary = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-joint-moped-baryon-run/des-y3-2pt-map3-all.txt', to_mcsamples=True)
s_joint_moped_auto_bary = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-joint-moped-baryon-run/des-y3-2pt-map3-auto.txt', to_mcsamples=True)
s_joint_moped_auto_334_344_bary = utils.read_cosmosis_mcmc_des_chain('../output/test-desy3c-joint-moped-baryon-run/des-y3-2pt-map3-auto-334-344.txt', to_mcsamples=True)
```

## Plotting

```python
g = plots.get_single_plotter(ratio=1, width_inch=6)
g.triangle_plot([s_joint_moped, s_joint_moped_bary], ['om', 's8', 'sig8', 'a1'], filled=False, \
    legend_labels=['joint', 'joint (baryon)'], legend_loc='upper right', 
    contour_colors=['gray', 'red'],
    markers={'om': 0.29, 's8':0.759, 'sig8': 0.772, 'a1':0.7})
plt.show()
```

```python
g = plots.get_single_plotter(ratio=1, width_inch=6)
g.triangle_plot([s_joint_moped_auto, s_joint_moped_auto_bary], ['om', 's8', 'sig8', 'a1'], filled=False, \
    legend_labels=['joint [auto]', 'joint (baryon) [auto]'], legend_loc='upper right', 
    contour_colors=['gray', 'red'],
    markers={'om': 0.29, 's8':0.759, 'sig8': 0.772, 'a1':0.7})
plt.show()
```
