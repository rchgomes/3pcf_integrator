```python
def create_dicts_one_angle(file):
    all_data = {'d2': [], 'Gamma0': [], 'Gamma1': [], 'Gamma2': [], 'Gamma3': [], 'Gamma_ttt': [], 'Gamma_xxx': [], 'Gamma_txx': []}
    for i in range(len(file)):
        all_data['d2'].append(file[i][8])
        all_data['Gamma0'].append(-file[i][0])
        all_data['Gamma1'].append(-file[i][2])
        all_data['Gamma2'].append(-file[i][4])
        all_data['Gamma3'].append(-file[i][6])
        all_data['Gamma_ttt'].append(-(file[i][0]+file[i][2]+file[i][4]+file[i][6])/4)
        all_data['Gamma_xxx'].append(-(file[i][1]+file[i][3]+file[i][5]+file[i][7])/4)
        all_data['Gamma_txx'].append(-(-file[i][1]-file[i][3]+file[i][5]+file[i][7])/4)
    return(all_data)
```
