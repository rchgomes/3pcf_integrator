```python
import sys
sys.path.append('../')
import threepoint
```

```python
tps1 = threepoint.ThreePointDataClass.from_fits('../data/dv/sim_map3-NLA-cosmoCosmoG.fits')
tps2 = threepoint.ThreePointDataClass.from_fits('../data/dv/sim_map3-NLA-cosmoCosmoG-phibin005.fits')
tpCG = threepoint.ThreePointDataClass.from_fits('../data/dv/map3_dv_COSMOGRID_19Apr24.fits')
```

```python
tps1.get_signal()
```

```python
plt.figure(figsize=(20, 5))

t = tps1.get_t_bin()[0,:]
b = np.arange(t.size)
plt.yscale('log')
plt.plot(b, t**2*tps1.get_signal(), marker='.')
plt.plot(b, t**2*tps2.get_signal(), marker='.')

std = tpCG.get_std()
plt.errorbar(b, t**2*tpCG.get_signal(), yerr=t**2*std, fmt='o')

plt.show()
```

```python
plt.figure(figsize=(20, 5))

t = tps1.get_t_bin()[0,:]
b = np.arange(t.size)
n = 1
base = tps1.get_signal()
tar = tps2.get_signal()
std = tpCG.get_std()/400**0.5
plt.errorbar(b, (tar/base), (std/base), fmt='o', color='gray', alpha=0.3)

for _b in b[3::4]:
    plt.axvline(_b+0.5, color='k', lw=0.6)

# plt.axhline((0.8/0.673)**-6)
# plt.axhline((0.7/0.673)**-6)
# plt.axhline((0.75/0.673)**-6)

plt.ylim(0, 3)
plt.axhline(1.0, color='k')
plt.ylabel('ratio = map3(sim)/map3(theory)', fontsize=15)
plt.xlabel('data id')

plt.show()
```
