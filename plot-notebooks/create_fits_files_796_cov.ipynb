```python
import sympy as sp

# Define the symbols
nsim, nd, np_sym = sp.symbols('nsim nd np', positive=True)

# Define the factor
factor = (1 + ((nd - np_sym) * (nsim - nd - 2)) / ((nsim - nd - 1) * (nsim - nd - 4)))**(-1)

# Compute the limit as nsim goes to infinity
limit_factor = sp.limit(factor, nsim, sp.oo)

print("The limit for nsim >> nd is:", limit_factor)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Fixed parameters
nsim = 500
nd = 80

# Define a range for np. Adjust the range as needed.
np_vals = np.linspace(2, 30, 400)

# Compute the factor for each np
factor_vals = (1 + ((nd - np_vals) * (nsim - nd - 2)) / ((nsim - nd - 1) * (nsim - nd - 4)))**(-1)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(np_vals, factor_vals, label='Factor')
plt.xlabel('np')
plt.ylabel('Factor Value')
plt.title('Plot of Factor vs np (nsim = 2000, nd = 100)')
plt.grid(True)
plt.legend()
plt.show()
```
