```python
#view the params and sigmas (2pt)
params = ['om', 's8', 'h0', 'cosmological_parameters--n_s', 'cosmological_parameters--omega_b', 'cosmological_parameters--mnu', 'a1']

for param in params:
    v1 = desy3_2ptonly.confidence(param, 0.16, upper=False)
    v2 = desy3_2ptonly.confidence(param, 0.16, upper=True)
    
    # Compute the peak (mean)
    peak = desy3_2ptonly.getMeans()
    
    print(f"Parameter: {param}")
    print(f"Peak: {peak}")
    print(f"1-sigma Lower: {v1}")
    print(f"1-sigma Upper: {v2}")
```
