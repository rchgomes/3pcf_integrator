```python
from astropy.io import fits
import numpy as np

# Paths to the input and output FITS files
input_fits_path = '../data/dv/sim_2pt-NLA-cosmoDESY3.fits'
output_fits_path = '../data/dv/blind-metacal-2pt-DESY3.fits'

# New arrays for the 'VALUE' fields in 'xip' and 'xim'
new_xip_values = np.load('../data/dv/xip_2pt_blind.npy')  # Replace with your new data array for 'xip' VALUE
new_xim_values = np.load('../data/dv/xim_2pt_blind.npy')  # Replace with your new data array for 'xim' VALUE

# Open the FITS file
with fits.open(input_fits_path, mode='update') as hdul:
    # Update the 'VALUE' field in HDU[2] for 'xip'
    hdul[2].data['VALUE'] = new_xip_values
    
    # Update the 'VALUE' field in HDU[3] for 'xim'
    hdul[3].data['VALUE'] = new_xim_values
    
    # Save changes to a new FITS file
    hdul.writeto(output_fits_path, overwrite=True)

print(f"Updated FITS file saved to {output_fits_path}")
```
