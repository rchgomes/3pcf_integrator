```python
import numpy as np
```

```python
#[cosmological_parameters]
omega_m = 0.290
S_8 = 0.759
h0 = 0.69
omega_b = 0.048
n_s = 0.97

#[shear_calibration_parameters]
m1 = 0.0
m2 = 0.0
m3 = 0.0
m4 = 0.0

#[wl_photoz_errors]
bias_1 = 0.0
bias_2 = 0.0
bias_3 = 0.0
bias_4 = 0.0

#[intrinsic_alignment_parameters]
A1      = 0.7
alpha1  = -1.7
```

```python
#all params
params = [omega_m, S_8, h0, omega_b, n_s, m1, m2, m3, m4, bias_1, bias_2, bias_3, bias_4, A1, alpha1]
deltas = 0.01*np.ones(shape=(15))
deltas[13] = 0.1
deltas[14] = 0.1
```

```python
shifts = np.zeros(shape=(4*len(params),len(params)))
```

```python
for i in range(len(params)):
    shifts[i*4] = params - 2*deltas[i]
    shifts[i*4+1] = params - deltas[i]
    shifts[i*4+2] = params + deltas[i]
    shifts[i*4+3] = params + 2*deltas[i]
```

```python
# Define the header
header = "# cosmological_parameters--omega_m cosmological_parameters--s_8 cosmological_parameters--h0 cosmological_parameters--omega_b cosmological_parameters--n_s shear_calibration_parameters--m1 shear_calibration_parameters--m2 shear_calibration_parameters--m3 shear_calibration_parameters--m4 wl_photoz_errors--bias_1 wl_photoz_errors--bias_2 wl_photoz_errors--bias_3 wl_photoz_errors--bias_4 intrinsic_alignment_parameters--a1 intrinsic_alignment_parameters--alpha1"

# Use np.savetxt to save the array to a text file with the specified header
np.savetxt("shifted_params.txt", shifts, header=header)
```
